# Training YOLOv8 SPBU (Lokal)
Jalankan cell di bawah ini berurutan untuk melatih model menggunakan GPU lokal.

```python
from ultralytics import YOLO
import torch

print('CUDA Tersedia:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU Device:', torch.cuda.get_device_name(0))
```

```python
import os
import yaml

dataset_path = os.path.abspath('dataset_spbu')
yaml_path = os.path.join(dataset_path, 'data.yaml')

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

data['path'] = dataset_path
data['train'] = 'images/train'
data['val'] = 'images/val'

with open(yaml_path, 'w') as f:
    yaml.dump(data, f)

print(f'File data.yaml sudah di-update dengan absolute path:\n{dataset_path}')
```

```python
model = YOLO('yolov8n.pt')

# Mulai Training
# workers=2 sangat penting di Windows untuk mencegah crash Memory
results = model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    name='spbu_model_lokal'
)
print('\n\nTRAINING SELESAI! Model terbaik tersimpan di runs/detect/spbu_model_lokal/weights/best.pt')
```
