```python
from ultralytics import YOLO
from pytubefix import YouTube
import cv2

# Panggil model terbaik yang baru saja selesai di-training
model = YOLO('runs/detect/spbu_model_lokal/weights/best.pt')

# Masukkan Link YouTube CCTV atau Lalu Lintas apa saja di sini
video_youtube = "https://youtu.be/f7o2ezFldZo?si=rYKYtQGISKqzY2cc"

# Ekstrak link MP4-nya secara manual menggunakan pytubefix
yt = YouTube(video_youtube)
stream_url = yt.streams.filter(file_extension="mp4").get_highest_resolution().url

# Kita loop secara manual agar tidak crash di Jupyter Windows
results = model.predict(source=stream_url, conf=0.5, stream=True, show=False)

print("Memulai video... Tekan 'q' pada keyboard untuk berhenti.")
try:
    for r in results:
        frame = r.plot()
        cv2.imshow("Video Deteksi", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
except KeyboardInterrupt:
    print("Dihentikan manual oleh user.")
finally:
    cv2.destroyAllWindows()
```

```python
import cv2
import glob
import random
from ultralytics import YOLO

# 1. Panggil model andalan kita
model = YOLO('runs/detect/spbu_model_lokal/weights/best.pt')

# 2. Ambil semua gambar asli dari folder dataset SPBU kita
daftar_gambar = glob.glob('dataset_spbu/images/val/*.jpg')

# Acak urutannya supaya yang dites beda-beda tiap kali dijalankan
random.shuffle(daftar_gambar)

print("Memulai tes... (Akan muncul pop-up window)")

# 3. Kita tes 15 gambar acak satu per satu
for img_path in daftar_gambar[:15]:
    # AI menebak gambar
    results = model.predict(source=img_path, conf=0.5)
    
    # Ambil hasil gambar yang sudah dikotaki oleh AI
    gambar_hasil = results[0].plot()
    
    # Tampilkan di layar
    cv2.imshow("Hasil Tes SPBU", gambar_hasil)
    
    # Jeda: Gambar akan berhenti sampai kamu menekan tombol apa saja (Spasi/Enter) di keyboard!
    cv2.waitKey(0)

# Tutup semua window kalau sudah selesai 15 gambar
cv2.destroyAllWindows()
print("Tes selesai!")
```
